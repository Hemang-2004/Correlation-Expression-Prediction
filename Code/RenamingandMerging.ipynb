```python
import pandas as pd

# Load the GEE satellite data 
file_path = '../Analysis/GEE-Dataset-Cleaned.csv'
df_gee = pd.read_csv(file_path)

# Mapping dictionary 
rename_map = {
    
    'Mahadevepura': 'Mahadevpura',
    
    'Agara': 'Agaram',
    'Allasandra': 'Allalasandra',
    'Bhoganahalli': 'Bhoganhalli',
    'Chellakere': 'Chelekere',
    'Devarabisanahalli': 'Devarabeesanahalli',
    'Gangashetti': 'Gangashetty',
    'Nagawara': 'Nagavara',
    'Panattur': 'Panathur',
    'Parapana': 'Parappana Agrahara',
    'Sarraki': 'Sarakki',
    'Sowl': 'Soulkere',
    'Vibhutipura': 'Vibhuthipura',
    'Vijinapura': 'Vijanapura',
    'Arekere': 'Arakere',
    'Thubarahalli': 'Tubarahalli',
    'Mallathahalli': 'Mallathahalli',
    
    'Gangashetti / Devasandra': 'Devasandra',
    'Kembathahali': 'Kembhatahali',
    'Nallurahallikere': 'Nallurahalli',
    'Doraikere': 'Uttarahalli Doraikere',
}

def rename_lakes(name):
    if pd.isna(name):
        return name
    
    # Clean string
    name_str = str(name).strip().strip(',')
    
    # Apply the mapping
    return rename_map.get(name_str, name_str)

# 1. Update the LAKE_NAME column
df_gee['LAKE_NAME'] = df_gee['LAKE_NAME'].apply(rename_lakes)

df_gee['LAKE_NAME'] = df_gee['LAKE_NAME'].str.title().str.strip()

# 3. Save and overwrite
df_gee.to_csv('GEE-Dataset-Cleaned_Final.csv', index=False)

print("Renaming complete")

# Verification check
check_list = ['Mahadevpura', 'Agaram', 'Allalasandra']
print("\nVerification check:")
found = df_gee[df_gee['LAKE_NAME'].isin(check_list)]['LAKE_NAME'].unique()
print(found)
```

```python
import pandas as pd

# Load the two datasets
# GEE-Dataset-Cleaned_Final.csv (Satellite Indices)
# Master_Dataset_Final_Sorted.csv (Ground Truth Monitoring)
df_gee = pd.read_csv('GEE-Dataset-Cleaned_Final.csv')
df_master = pd.read_csv('Master_Dataset_Final_Sorted.csv')

# Merging
# This keeps only rows where LAKE_NAME AND DATE exist in both files.
merged_df = pd.merge(df_gee, df_master, on=['LAKE_NAME', 'DATE'], how='inner')


merged_df.to_csv('Merged_Lake_Water_Quality_Master.csv', index=False)


print("Merge Complete!")
print(f"Rows in GEE file: {len(df_gee)}")
print(f"Rows in Master file: {len(df_master)}")
print(f"Rows in Merged file: {len(merged_df)}")

print("\nPreview of the merged data (First 5 rows):")
print(merged_df.head())

# Check for any specific lake to see if the match worked
if not merged_df.empty:
    sample_lake = merged_df['LAKE_NAME'].iloc[0]
    print(f"\nSuccessful match found for: {sample_lake}")
else:
    print("\nWarning: No matches found.")
```
